```python
import os

dataset_path = r"C:\Users\DELL\Documents\EcoSort-AI\dataset\waste_dataset\standardized_256"

if os.path.exists(dataset_path):
    print("Dataset folder found successfully!")
else:
    print("Dataset folder not found. Check the path.")
```

```python
# Get all waste categories from the dataset folder

categories = os.listdir(dataset_path)

# Sort categories alphabetically
categories.sort()

print("Waste Categories:")
print("-----------------")

for category in categories:
    print(category)

print("\nTotal number of categories:", len(categories))
```

```python
# Count images in each waste category

image_extensions = ('.png', '.jpg', '.jpeg')

total_images = 0

for category in categories:

    category_path = os.path.join(dataset_path, category)

    images = [
        file for file in os.listdir(category_path)
        if file.lower().endswith(image_extensions)
    ]

    image_count = len(images)
    total_images += image_count

    print(f"{category}: {image_count} images")

print("---------------------------")
print("Total images:", total_images)
```

```python
import os
import matplotlib.pyplot as plt

# Define dataset location
dataset_path = r"C:\Users\DELL\Documents\EcoSort-AI\dataset\waste_dataset\standardized_256"

# Get category names
categories = sorted([
    folder for folder in os.listdir(dataset_path)
    if os.path.isdir(os.path.join(dataset_path, folder))
])

# Count images in each category
image_counts = []

for category in categories:
    category_path = os.path.join(dataset_path, category)

    count = len([
        file for file in os.listdir(category_path)
        if file.lower().endswith(('.png', '.jpg', '.jpeg'))
    ])

    image_counts.append(count)

# Plot dataset distribution
plt.figure(figsize=(12, 6))

plt.bar(categories, image_counts)

plt.title("EcoSort AI - Dataset Distribution")
plt.xlabel("Waste Categories")
plt.ylabel("Number of Images")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os
import random

# Display one sample image from each category

plt.figure(figsize=(15, 10))

for index, category in enumerate(categories):

    category_path = os.path.join(dataset_path, category)

    # Get image filenames
    images = [
        file for file in os.listdir(category_path)
        if file.lower().endswith(('.png', '.jpg', '.jpeg'))
    ]

    # Select one random image
    random_image = random.choice(images)

    # Complete image path
    image_path = os.path.join(category_path, random_image)

    # Open image
    img = Image.open(image_path)

    # Display image
    plt.subplot(2, 5, index + 1)
    plt.imshow(img)
    plt.title(category.capitalize())
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import numpy as np
import os

# Select one sample image
category = "battery"

category_path = os.path.join(dataset_path, category)

images = [
    file for file in os.listdir(category_path)
    if file.lower().endswith(('.png', '.jpg', '.jpeg'))
]

image_path = os.path.join(category_path, images[0])

# Open the image
img = Image.open(image_path)

# Convert image into a NumPy array
image_array = np.array(img)

# Normalize pixel values
normalized_image = image_array / 255.0

# Display information
print("Original Image Shape:", image_array.shape)
print("Original Minimum Pixel Value:", image_array.min())
print("Original Maximum Pixel Value:", image_array.max())

print("\nNormalized Image Shape:", normalized_image.shape)
print("Normalized Minimum Pixel Value:", normalized_image.min())
print("Normalized Maximum Pixel Value:", normalized_image.max())
```
